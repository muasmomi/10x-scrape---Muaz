```python
header = ['SKU', 'Link']
with open(fr"0212 DON-products.csv", 'a', newline='', encoding='utf-8-sig') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    df1 = pd.read_csv(fr"0211 DON PL.csv", dtype = str)

    # display(df1)
    for index, row in df1.iterrows():
        try:
            sku = row['Item'].strip()
            browser.get(f'https://www.autovaux.co.uk/catalogsearch/result/?q={sku}')
            time.sleep(2.5)
        
            title = browser.find_element(by=By.CSS_SELECTOR, value='h1.page-title span').get_attribute('innerText').strip()
            sku0 = browser.find_element(by=By.CSS_SELECTOR, value='[itemprop="sku"]').get_attribute('innerText').strip()
            vendor = title.split(' ')[0].strip()

            try:
                descs_el = browser.find_elements(by=By.CSS_SELECTOR, value='#description li')
                descs = [i.get_attribute('innerText').strip() for i in descs_el]
            except:
                descs = []

            try:
                features = browser.find_element(by=By.CSS_SELECTOR, value='#specification table').get_attribute('innerHTML')
                features = re.sub(r'<(\w+)(\s+[^>]*?)?>', r'<\1>', features)
                features = re.sub(r'\s+', ' ', features)
            except:
                features = ''

            images = browser.find_elements(by=By.CSS_SELECTOR, value='.fotorama__stage__frame .fotorama__img')
            image_list = []
            for i in images:
                image = i.get_attribute('src')
                image_list.append(image)

            if sku == sku0:                        
                writer.writerow([sku, title, vendor, descs, features, image_list])
                print([sku, title, vendor, descs, features, image_list])
            
        except Exception as e:
            print(f'{str(e).splitlines()[0]}')
```

```python
csv_path = r"0212 DON-products_2.csv"
scrape_df = pd.read_csv(csv_path).fillna('')

final_data = []

for index0, row in scrape_df.iterrows():

    vendor = 'DON'
    sku = row['sku']
    title = f'{vendor} ' + f'{sku} ' +row['title'].replace(f'{vendor}', '').replace(f'{sku} ', '').strip()
    sfeatures = row['features']
    weight = row['weight']
    ean = row['barcode']
    cost = row['cost']

    if sfeatures == '':
        features = ''
    else:
        features = f'<p>&nbsp;</p><h4>Features</h4><table>{sfeatures}</table>'

    descs = ast.literal_eval(row['descs'])
    compatibilities = [f"<li>{item.replace('Fits ', '')}</li>" for item in descs if 'fits' in item.lower()]

    if compatibilities:
        compatibility = f'<ul>{"".join(compatibilities)}</ul>'
    else:
        compatibility = ''
    
    remove_keywords = ["Supplied By", "Part Numbers", "Fits", "Guarantee", "Supplied by"]

    desc0 = [
        f"<p>{item}</p>" for item in descs
        if not any(keyword in item for keyword in remove_keywords)
    ]

    if desc0:
        desc = ''.join(desc0)
    else:     
        desc = f'<p>This is {title}</p>'

    images = ast.literal_eval(row['images'])

    handle = (re.sub(r'[^a-zA-Z0-9\n\.]', '-', title).replace(".", "-").replace("---", "-").replace("--", "-")).lower()
    if handle.endswith('-'):
        handle = handle[:-1]

    def description(desc, features, compatibility, sku):
        return f"""<h4><strong>Description</strong></h4>{desc}
        {features}
        <p>&nbsp;</p>
        <h4>Compatibility</h4>{compatibility}<p>Feel free to contact us at info@mlperformance.co.uk should you wish to double check!</p>
        <p>&nbsp;</p>
        <h4>Compatibility Check</h4><p>To ensure the part(s) you have ordered fits your vehicle, we run a compatibility check prior to dispatch. We can do this either using your registration number (UK) or the last 7 digits of your VIN. Simply enter your car details prior to checkout.</p>
        <p>&nbsp;</p>
        <h4>Part Number</h4><p>{vendor.upper()}-{sku}</p>
        <p>&nbsp;</p>
        <h4>More Information</h4><p><strong>Manufactured by</strong></p><p>{vendor}</p>"""

    for index, image in enumerate(images, start=1):
        info = {}
        
        info['Handle'] = handle
        info['Title'] = title
        info['Body (HTML)'] = description(desc, features, compatibility, sku).replace("\n", "")
        info['Vendor'] = vendor
        info['Standardized Product Type'] = None
        info['Custom Product Type'] = None
        info['Tags'] = f"Uploaded by_Muazzim, 2026_New Products, MLP Discount 2026, Brand_{vendor}, Product Type_"
        info['Published'] = "TRUE"
        info['Option1 Name'] = 'Title'
        info['Option1 Value'] = 'Default Title'
        info['Option2 Name'] = None
        info['Option2 Value'] = None
        info['Option3 Name'] = None
        info['Option3 Value'] = None

        info['Variant SKU'] = f'{vendor.upper()}-{sku}'
        try:
            info['Variant Grams'] = round(float(weight)*1000, 2)
        except:
            info['Variant Grams'] = ''
            
        info['Variant Inventory Tracker'] = "shopify"
        info['Variant Inventory Policy'] = 'continue'
        info['Variant Fulfillment Service'] = 'manual'
        info['Variant Price'] = None
        info['Variant Compare At Price'] = ''
        info['Variant Requires Shipping'] = 'TRUE'
        info['Variant Taxable'] = 'TRUE'
        info['Variant Barcode'] = ean

        # For each image, create a new entry
        info['Image Src'] = image
        info['Image Position'] = index
        info['Image Alt Text'] = title
        info['Gift Card'] = None
        info['SEO Title'] = title
        info['SEO Description'] = 'Get ' + title + ' for your car to get your desired looks and performance from ML Performance at the lowest price with FREE UK shipping & next day delivery on in stock items. Very cheap prices & good service.'
        info['Google Shopping / Google Product Category'] = None
        info['Google Shopping / Gender'] = None
        info['Google Shopping / Age Group'] = None
        info['Google Shopping / MPN'] = sku
        info['Google Shopping / AdWords Grouping'] = None
        info['Google Shopping / AdWords Labels'] = None
        info['Google Shopping / Condition'] = 'new'
        info['Google Shopping / Custom Product'] = None
        info['Google Shopping / Custom Label 0'] = None
        info['Google Shopping / Custom Label 1'] = None
        info['Google Shopping / Custom Label 2'] = None
        info['Google Shopping / Custom Label 3'] = None
        info['Google Shopping / Custom Label 4'] = None
        info['Variant Image'] = None
        info['Variant Weight Unit'] = 'kg'
        info['Variant Tax Code'] = 8708949900
        info['Cost per item'] = cost
        info['Margins'] = None
        info['Price / International'] = None
        info['Compare At Price / International'] = None
        info['Status'] = 'active'

        final_data.append(info)

final_df = pd.DataFrame(final_data)

output_path = os.path.join("0219 DON.csv")
final_df.to_csv(output_path, index=False)

print('File saved and moved to desired folder')
```
